### Record Granularity — Documentation Review

**Correction supported by documentation:** the initial interpretation that every row is an individual journal voucher item was too narrow. The published structure accommodates **journal-voucher entries and accounting/control summary records**. A row should be treated as an accounting record whose underlying detail depends on its type, not automatically as one original payment, obligation, or cash movement.

The [official XML dictionary](https://donnees-data.tpsgc-pwgsc.gc.ca/ba1/SGBT-BCMS-dd.xml) defines the four relevant fields as follows (paraphrased; XML heading IDs identify the definitions):

| Field | Documented meaning | Consequence for granularity |
|---|---|---|
| Accounting Control Number (`h9`) | Identifies an accounting transaction prepared for general-ledger posting; several transactions may be grouped under it | Repetition is allowed by the documented meaning; it is not a row identifier by itself |
| Journal Voucher Identifier (`h76`) | Optional voucher reference assigned by the Common Departmental Financial System | Absence does not prove a defective record or establish its type |
| Journal Voucher Item Identifier (`h78`) | Sequence number for an individual general-ledger entry within a voucher | Identifies voucher-entry detail, not necessarily one underlying business event |
| Accounting Summary Transaction Line Count (`h81`) | Number of detail transactions represented by a grouping; may be blank for voucher transactions | A summary can represent multiple detail transactions; this is not simply a CSV-row count |

The [dataset-specific supporting documentation](https://donnees-data.tpsgc-pwgsc.gc.ca/ba1/tcrg-rgat/soutien-support-eng.html) explicitly distinguishes journal-voucher transactions from control data in its Department Number instructions. This confirms that a uniform voucher-only interpretation is inappropriate.

**Existing local evidence:** the displayed sample already contains summary counts greater than one. That supports grouped detail in the local extract without any new analysis. The documentation does not provide a complete row-classification algorithm or a mapping from every local row to its underlying source events. Those remain unverified. The observed composite-key result is unchanged and does not establish uniform business granularity.

```python
df.duplicated(
    subset=[
        "Accounting-Control-Number-Numéro-contrôle-comptable",
        "Journal-Voucher-Item-Identifier-Identificateur-de-l'item-de-la-pièce-de-journal",
    ]
).sum()
```

### Temporary Numeric Profiling — Explicitly Authorized

The user authorized temporary numeric parsing for profiling only. The original DataFrame and CSV remain unchanged. Parsing uses these explicit conventions:

- A decimal point separates two fractional digits. Commas in validated three-digit or Indian-style two/three-digit groups are interpreted as grouping separators; surrounding spaces and `$` are formatting, not currency identification.
- Matched parentheses are interpreted as negative amounts for this profile. This is an accounting-format assumption, not evidence of an error or verified source semantics.
- Dollar-dash strings remain **unresolved**. They are represented as unavailable only in the temporary numeric view and excluded from descriptive numeric statistics; they are not imputed as zero and original strings remain intact.
- Unexpected formats cause an error instead of silently discarding values. Units Sold is parsed as a magnitude for exploration only; its dollar formatting and fractional units still need business clarification.

Statistics therefore cover 700 numeric values for most measures, **647 for Discounts**, and **695 for Profit**. Floating-point arithmetic here is descriptive profiling, not an approved operational monetary representation.

```python
company_numeric_profile = pd.DataFrame(index=company_financials.index)
company_amount_pattern = (
    r"\s*\$\s*(?P<open>\()?\s*"
    r"(?P<amount>(?:\d{1,3}(?:,\d{3})+|\d{1,2}(?:,\d{2})*,\d{3}|\d+)\.\d{2})"
    r"\s*(?P<close>\))?\s*"
)
for company_column in company_measures:
    company_numeric_values = []
    for company_raw_value in company_financials[company_column]:
        if re.fullmatch(r"\s*\$-\s*", company_raw_value):
            company_numeric_values.append(float("nan"))
            continue
        company_match = re.fullmatch(company_amount_pattern, company_raw_value)
        if company_match is None or bool(company_match["open"]) != bool(company_match["close"]):
            raise ValueError(f"Unrecognized profiling format: {company_raw_value!r}")
        company_magnitude = float(company_match["amount"].replace(",", ""))
        company_numeric_values.append(-company_magnitude if company_match["open"] else company_magnitude)
    company_numeric_profile[company_column] = company_numeric_values
print("Numeric descriptive statistics; dollar-dash placeholders excluded:")
print(company_numeric_profile.describe().T.to_string(float_format=lambda value: f"{value:,.2f}"))
print("Distribution profile under the stated parsing conventions:")
print(pd.DataFrame({
    "skewness": company_numeric_profile.skew(),
    "negative_count": (company_numeric_profile < 0).sum(),
    "explicit_numeric_zero_count": (company_numeric_profile == 0).sum(),
    "unresolved_placeholder_count": company_numeric_profile.isna().sum(),
}).to_string(float_format=lambda value: f"{value:,.3f}"))
```
