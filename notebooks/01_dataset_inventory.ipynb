## 10. Financial Measure Analysis

The main monetary attribute is analyzed to understand its statistical distribution.

The objective is to identify the scale, variability, central tendency, and extreme values of accounting transaction amounts.

Extreme values are not automatically considered errors because large financial transactions may be legitimate.

```python
df["Journal-Voucher-Item-Amount-Montant-de-l'item-de-la-pièce-de-journal"].describe()
```

### Preliminary Observations

The transaction amount distribution is highly right-skewed.

The median transaction amount is approximately 1,767.72, while the mean is substantially higher due to a relatively small number of extremely large accounting entries.

The maximum observed transaction amount is approximately 39.56 billion.

This suggests the presence of strong outliers or legitimately large accounting transactions. These observations should not be removed automatically and will require further investigation during the EDA and anomaly-detection stages.

Because of the highly skewed distribution, the median may provide a more representative measure of a typical transaction than the arithmetic mean.





### Debit and Credit Comparison

Transaction amounts are also analyzed separately by debit and credit classification.

This provides a clearer understanding of the monetary behavior of the two accounting entry types.

```python
df.groupby("Credit/Debit-Code-Code-Crédit/Débit")[
    "Journal-Voucher-Item-Amount-Montant-de-l'item-de-la-pièce-de-journal"
].agg(["count", "sum", "mean", "median", "min", "max"])
```

### Debit and Credit Observations

Debit entries are considerably more frequent than credit entries in the dataset.

Although the median transaction amounts for debit and credit entries are relatively similar, debit entries contain substantially larger extreme values.

The total debit amount is also significantly higher than the total credit amount in this dataset.

These differences should not be interpreted directly as expenses versus revenues, because debit and credit classifications represent accounting entry directions rather than business income and expense categories.

The result suggests that debit and credit transactions may need to be analyzed separately during anomaly detection and financial analysis.

## 11. Duplicates and Potential Keys

This section evaluates complete duplicate rows and investigates columns that may function as identifiers.

A column with high cardinality is not automatically unique.

In accounting datasets, a control number may identify a journal entry containing multiple individual items, which means that a combination of attributes may be required to uniquely identify each record.

```python
df.duplicated().sum()
```

```python
df["Accounting-Control-Number-Numéro-contrôle-comptable"].duplicated().sum()
```

### Duplicate Analysis

No complete duplicate rows were identified in the dataset.

However, the Accounting Control Number contains 148 duplicated occurrences, confirming that this attribute cannot uniquely identify individual records by itself.

This result motivates the investigation of a composite key.

### Key Investigation

If the accounting control number is repeated across multiple records, the dataset may require a composite key.

Possible combinations of control numbers, journal voucher identifiers, and journal voucher item identifiers should be investigated before defining the final database schema.

### Composite Key Investigation

The Accounting Control Number is not unique by itself.

This test evaluates whether combining the Accounting Control Number with the Journal Voucher Item Identifier produces a unique identifier for each record.

```python
df.duplicated(
    subset=[
        "Accounting-Control-Number-Numéro-contrôle-comptable",
        "Journal-Voucher-Item-Identifier-Identificateur-de-l'item-de-la-pièce-de-journal",
    ]
).sum()
```

### Composite Key Result

The combination of Accounting Control Number and Journal Voucher Item Identifier produced zero duplicate records.

This indicates that the two attributes together uniquely identify each row in the current dataset.

Therefore, this combination is a strong candidate for a composite key.

The final database key should still be confirmed during the schema-design stage and validated against the official dataset documentation.
